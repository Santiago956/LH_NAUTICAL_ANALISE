# Análise Exploratória de Dados (EDA) e Verificação de Nulos

Este notebook visa explorar a nossa `Fato de Vendas` já tratada e realizar verificações compreensivas de qualidade (especialmente dados ausentes).

```python
import duckdb
import pandas as pd

# Carregando a Fato de Vendas
con = duckdb.connect()
df = con.execute("SELECT * FROM '../data/processed/fato_vendas.parquet'").df()
print(f"Total de Linhas: {len(df):,}")
df.head()
```

```python
# 1. Verificação Compreensiva de Nulos (Missing Values)
null_counts = df.isnull().sum()
null_pct = (null_counts / len(df)) * 100

nulos = pd.DataFrame({'Nulos': null_counts, '%': null_pct})
nulos_encontrados = nulos[nulos['Nulos'] > 0].sort_values('Nulos', ascending=False)

if len(nulos_encontrados) > 0:
    display(nulos_encontrados)
else:
    print("Nenhum valor nulo encontrado na tabela Fato!")
```

```python
# 2. Estatísticas Descritivas das Variáveis Financeiras
df[['item_revenue', 'total_cost', 'gross_margin']].describe().round(2)
```

```python
# 3. Distribuição de Vendas por Canal (E-commerce vs Lojas Físicas)
df.groupby('channel')['item_revenue'].sum().sort_values(ascending=False).apply(lambda x: f"R$ {x:,.2f}")
```
